```python
from pyspark.sql import SparkSession
import pyspark.sql.functions as F
from pyspark.sql.window import Window

spark = SparkSession.builder.appName("WindowFct").getOrCreate()
```

```python
data = [
    (1, 'Amit', 'IT', 60000),
    (2, 'Neha', 'IT', 75000),
    (3, 'Raj', 'IT', 60000),
    (4, 'Priya', 'HR', 50000),
    (5, 'Karan', 'HR', 50000),
    (6, 'Sneha', 'HR', 70000),
    (7, 'Vikas', 'Sales', 45000),
    (8, 'Rohit', 'Sales', 55000),
]
cols = [
    "emp_id",
    "name",
    "department",
    "salary",
]
df = spark.createDataFrame(data, cols)
df.show()
```

```python
win = Window.partitionBy("department").orderBy(F.col("salary").desc())
```

```python
res = (
    df
    .withColumn("row_num",
        F.row_number().over(win)
    )
    .withColumn("rank",
        F.rank().over(win)
    )
    .withColumn("dense_rank",
        F.dense_rank().over(win)
    )
    .withColumn("pct_rank",
        F.percent_rank().over(win)
    )
)

res.show()
```

```python
from pyspark.sql import SparkSession
import pyspark.sql.functions as F
from pyspark.sql.window import Window

spark = SparkSession.builder.appName("WindowFct").getOrCreate()

data = [
    (1, 'Amit', 'IT', 60000),
    (2, 'Neha', 'IT', 75000),
    (3, 'Raj', 'IT', 60000),
    (4, 'Priya', 'HR', 50000),
    (5, 'Karan', 'HR', 50000),
    (6, 'Sneha', 'HR', 70000),
    (7, 'Vikas', 'Sales', 45000),
    (8, 'Rohit', 'Sales', 55000),
]
cols = [
    "emp_id",
    "name",
    "department",
    "salary",
]
df = spark.createDataFrame(data, cols)

# define the window
win = Window.partitionBy("department").orderBy(F.col("salary").desc())

res = (
    df
    .withColumn("row_num",    F.row_number().over(win))
    .withColumn("rank",       F.rank().over(win))
    .withColumn("dense_rank", F.dense_rank().over(win))
    .withColumn("pct_rank",   F.percent_rank().over(win))
)

res.show()
```
