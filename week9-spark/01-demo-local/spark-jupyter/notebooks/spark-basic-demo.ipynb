```python
from pyspark.sql import SparkSession

# Initialize Spark session
spark = SparkSession.builder.appName("VerifySpark").getOrCreate()
```

```python
spark, type(spark)
```

```python
# Sample data
data = [("Alice", 25), ("Bob", 30), ("Cathy", 28)]
columns = ["Name", "Age"]
```

```python
# Create DataFrame
df = spark.createDataFrame(data, columns)

# Display DataFrame
df.show()
```

```python
# Stop Spark session
spark.stop()
```

```python
with SparkSession.builder.appName("VerifySpark").getOrCreate() as session1:
    df1 = session1.createDataFrame(data, columns)
    print(df1.show())
    print(df1.collect())
    print(df1["Name"])
    
```
