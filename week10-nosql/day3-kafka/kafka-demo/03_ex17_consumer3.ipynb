```python
# Note: Run Consumer 3 in a separate cell or script
from kafka import KafkaConsumer
import time
consumer3 = KafkaConsumer(
    'partitioned-topic',
    bootstrap_servers="host.docker.internal:9093",
    group_id='test-group',
    auto_offset_reset='earliest'
)
```

```python
print("Consumer 3 started...")
for message in consumer3:
    print(f"Consumer 3, Partition {message.partition}: {message.value.decode('utf-8')}")
    time.sleep(0.5)
    #break
```

```python
consumer3.close()
```
